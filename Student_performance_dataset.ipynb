```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
path=r"/content/archive (17).zip"
data=pd.read_csv(path)
print(data.shape)

print(data.head())
```

```python
print(data.isnull().sum())

print(data.dtypes)
```

```python
print(data['Branch'])
```

```python
data['Branch']=data['Branch'].ffill()

# print(data['Branch'])

data['Branch']=data['Branch'].fillna('Missing')

print(data['Branch'])
```

```python
print('Duplicates:',data.duplicated().sum())
```

```python
sns.displot(data=data,y='Attendance(%)',x='Marks(%)')
plt.ylabel('Attendance in percentage')
plt.xlabel('Marks in percentage')
plt.title('Relationship of attendance and final marks')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=data, x='Course', y='Marks(%)')
plt.title('Distribution of Marks by Course')
plt.xlabel('Course')
plt.ylabel('Marks (%)')
plt.show()
```
